```python
from operator import index

import numpy as np
import pandas as pd
```

Series from lists

```python
#string
country=['india','pakistan','USA','nepal','srilanka']
pd.Series(country)
```

```python
#interger
marks=[10,92,42,14,3,1]
pd.Series(marks)
```

```python
#custom index
marks=[54,52,65,75,87]
subject=['maths','english','chemistry','science','physics']

pd.Series(marks,index=subject)
```

```python
#setting a  name
pd.Series(marks,index=subject,name='laxman')
```

Series from dict

```python
marks={
    'maths':34,
    'english':64,
    'chemistry':75,
    'science':64,
    'physics':87
}

marks_Series=pd.Series(marks,name='laxman')
marks_Series
```
